```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

fig, ax = plt.subplots(figsize=(12, 6))

ax.hist(df_customers['annual_revenue_potential'].dropna(), bins=25, edgecolor='black')

# Format x-axis values with thousands separators
ax.xaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))

ax.xaxis.set_major_locator(ticker.MaxNLocator(nbins=10))

ax.set_xlabel('Annual Revenue Potential')
ax.set_ylabel('Number of Customers')
ax.set_title('Distribution of Annual Revenue Potential')

plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.show()
```

```python
df_active_customer = df_customers[df_customers['account_status'] == "Active"]
```

```python
# Percentage of active clients, over total of 200 customers
int(df_active_customer['account_status'].count()) / 200 * 100
```

```python
freight_types = df_active_customer['primary_freight_type'].unique()
freight_types
```

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

for ax, freight_type in zip(axes, freight_types):
    data = df_active_customer.loc[
        df_active_customer['primary_freight_type'] == freight_type,
        'annual_revenue_potential'
    ].dropna()

    ax.boxplot(data, showmeans=True, meanline=True)
    ax.set_title(freight_type)
    ax.set_ylabel('Annual Revenue Potential ($)')
    ax.set_yticks(np.linspace(500_000, 5_000_000, 10))
    ax.grid(axis='y', linestyle='--', linewidth=0.5)

fig.suptitle('Annual Revenue Potential by freight type', fontsize=16)
plt.tight_layout()
plt.show()
```

### In the plot we clearly notice that the two most potential profitable markets for the logistic company customers portfolio are Consumer Goods and Food/Beverage

```python
df_fuel = pd.read_csv("./data/fuel_purchases.csv")
```

```python
df_fuel.info()
```

```python
df_fuel.head(10)
```

```python
df_fuel.set_index('purchase_date')['price_per_gallon'].resample('MS').max()
```

```python
df_fuel['purchase_date'] = pd.to_datetime(df_fuel['purchase_date'])

monthly_avg = (
    df_fuel
    .set_index('purchase_date')['price_per_gallon']
    .resample('MS') # month start
    .mean()
)


fig, ax = plt.subplots(figsize=(12, 6))

ax.plot(monthly_avg.index, monthly_avg.values, marker='o')

ax.set_title('Average Monthly Fuel Price')
ax.set_xlabel('Month')
ax.set_ylabel('Average Price per Gallon ($)')
ax.grid(True, linestyle='--', linewidth=0.5)

fig.autofmt_xdate()
plt.tight_layout()
plt.show()
```

```python
df_incidents = pd.read_csv("./data/safety_incidents.csv")
```

```python
df_incidents.info()
```

```python
df_incidents = df_incidents.dropna()
```

```python
df_incidents.head(10)
```

```python
incidents_merged = df_incidents.groupby('driver_id').size() \
    .reset_index(name='incident_count') \
    .sort_values('incident_count', ascending=False)\
    .merge(df_drivers)[['incident_count', 'years_experience']]
```

```python
methods = ['pearson', 'spearman', 'kendall']

for method in methods:
    corr = incidents_merged['incident_count'].corr(
        incidents_merged['years_experience'],
        method=method
    )
    print(f'{method.capitalize()}: {corr:.3f}')
```

### But correlation is not causation. There's might be a lot of other factors that determines this small positive correlation
